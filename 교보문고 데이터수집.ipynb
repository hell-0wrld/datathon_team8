```python
import requests
```

```python
# 요청을 보낼 URL
url = 'https://store.kyobobook.co.kr/api/gw/best/best-seller/total?page=1&per=200&period=003&bsslBksClstCode=A&ymw=202501'  # 원하는 API 주소로 바꿔주세요

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/115.0.0.0 Safari/537.36"
    )
}

response = requests.get(url, headers=headers)

# 응답 상태 코드 확인
if response.status_code == 200:
    # JSON 파싱
    meta = response.json()
    print(meta)
else:
    print(f"요청 실패: {response.status_code}")
```

```python
book_infos = [[book['cmdtName'], book['saleCmdtid']] for book in meta['data']['bestSeller']]
```

```python
import time
import random

review_dict = {}

# 책을 3개씩 끊어서 가져오기
for i in range(0, len(book_infos), 3):
    batch = book_infos[i:i+3]

    for book_info in batch:
        book_title = book_info[0]
        book_id = book_info[1]
        
        url = (
            f'https://product.kyobobook.co.kr/api/review/list?page=1&pageLimit=900'
            f'&reviewSort=001&revwPatrCode=002&saleCmdtids={book_id}'
            f'&webToonYsno=N&allYsno=N&revwSummeryYn=Y&saleCmdtid={book_id}'
        )

        headers = {
            'User-Agent': (
                'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
                'AppleWebKit/537.36 (KHTML, like Gecko) '
                'Chrome/115.0.0.0 Safari/537.36'
            ),
        }

        try:
            response = requests.get(url, headers=headers)
            if response.status_code == 200:
                review = response.json()
                review_dict[book_title] = review['data']['reviewList']
            else:
                print(f"[{book_title}] 요청 실패: {response.status_code}")
        except Exception as e:
            print(f"[{book_title}] 오류 발생: {e}")

    # 진행상황 출력 및 딜레이
    random_sec = random.uniform(3, 5)
    print(f"[{i+1}~{i+len(batch)}] 책 리뷰 수집 완료. 대기 {random_sec:.2f}초...")
    time.sleep(random_sec)
```

```python
merged_data = []

# 메타 정보에서 책 정보 추출
meta_books = meta['data']['bestSeller']

for book in meta_books:
    book_ID = book['saleCmdtid']
    book_title = book['cmdtName']
    rank = book.get('prstRnkn')
    overall_rating = book.get('buyRevwRvgr')
    review_count = book.get('buyRevwNumc')
    main_keywords = book.get('revwEmtnKywrName')  # 문자열일 수도 있고, 리스트일 수도 있음

    # 해당 책에 대한 리뷰가 있는 경우만 처리
    if book_title in review_dict:
        for review in review_dict[book_title]:
            merged_data.append({
                'rank': rank,
                'book_ID': book_ID,
                'book_title': book_title,
                'overall_rating': overall_rating,
                'review_count': review_count,
                'main_keywords': main_keywords,
                'review': review.get('revwCntt'),
                'keyword': review.get('revwEmtnKywrName'),
                'rating' : review.get('revwRvgr')
            })
```
