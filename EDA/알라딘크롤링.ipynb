```python
import requests

# 요청을 보낼 URL
url = 'https://store.kyobobook.co.kr/api/gw/best/best-seller/total?page=1&per=200&period=003&bsslBksClstCode=A&ymw=202501'  # 원하는 API 주소로 바꿔주세요

# GET 요청 보내기
response = requests.get(url)

# 응답 상태 코드 확인
if response.status_code == 200:
    # JSON 파싱
    data = response.json()
    print(data)
else:
    print(f"요청 실패: {response.status_code}")
```
