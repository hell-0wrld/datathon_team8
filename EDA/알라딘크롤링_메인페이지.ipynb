```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

import time 
import random

headers = {
    'User-Agent': (
    'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
    'AppleWebKit/537.36 (KHTML, like Gecko) '
    'Chrome/115.0.0.0 Safari/537.36'
    ),
    }


data_list = [] # 데이터를 담을 리스트
for year in range(2024, 2025):
    for month in range(1, 13):
        for week in range(1, 6):
          for page in range(1, 11):
            
            # 요청을 보낼 URL
            url = f'https://www.aladin.co.kr/shop/common/wbest.aspx?BranchType=1&CID=0&Year={year}&Month={month}&Week={week}&BestType=Bestseller&page={page}&SearchSubBarcode='
            # GET 요청 보내기
            response = requests.get(url, headers = headers)

            soup = BeautifulSoup(response.text, 'html.parser')


            book_containers = soup.select('.ss_book_box')
            for i in book_containers:
                book_data = {} # 각 책의 데이터를 담을 딕셔너리

                # 제목
                title = i.select_one('a.bo3')
                if title:
                    book_data['book_name'] = title.text
                else:
                    pass

                # 책 코드
                link = i.select_one('a.bo3')
                if link:
                    al_ID = link['href'].split('=')[1]
                    book_data['book_id'] = al_ID
                else:
                    pass

                # 작가
                auth = i.select_one('a[href^="/Search/wSearchResult.aspx?AuthorSearch="]')
                if auth:
                    book_data['auth_name'] = auth.text
                else:
                    pass

                # 작가 코드
                auth_link = i.select_one('a[href^="/Search/wSearchResult.aspx?AuthorSearch="]')
                if auth_link:
                    auth_ID = auth_link['href'].split('@')[1].split('&')[0]
                    book_data['auth_id'] = auth_ID
                else:
                    pass

                # 출판사, 날짜
                publi = i.select_one('a[href^="/search/wsearchresult.aspx?PublisherSearch="]')
                if publi:
                    book_data['publi_name'] = publi.text
                    date_info_raw = publi.next_sibling.strip()
                    date_info = date_info_raw.split(' ')[1] + ' ' + date_info_raw.split(' ')[2]
                    book_data['publi_date'] = date_info
                else:
                    pass

                # 랭크 날짜
                book_data['rank_date'] = f"{year}-{month:02d}-{week:02d}"


                # 가격
                selling_price_element = i.select_one('span.ss_p2')
                if selling_price_element:
                    # 앞거로 가져옴
                    original_price_span = selling_price_element.find_previous_sibling('span')
                    if original_price_span:
                        original_price = original_price_span.text.strip()
                        price = original_price.replace(',','')
                
                        # 정수화
                        price = int(price)
                        
                        book_data['price'] = price

                # 랭크
                rank_element = i.select_one('td div[style="text-align: center;"]')
                if rank_element:
                    rank_text = rank_element.text.strip()
                    # "51."과 같은 형태에서 "51"만 추출
                    rank = int(rank_text.replace('.', ''))
                    book_data['rank'] = rank
                else:
                    book_data['rank'] = None

                # 총 별점
                star = i.select_one('span.star_score')
                if star:
                    star = star.text
                    star = float(star)
                    book_data['star'] = star
                else:
                    book_data['star'] = None

                # 별점 수
                a = i.select_one('a[href*="CommentReview"]')
                if a:
                    a= a.text
                    a = int(a)
                    book_data['star_count'] = a
                else:
                    book_data['star_count'] = None

                data_list.append(book_data) # 추출된 책 데이터 추가



                

            # 각 페이지 요청 후 랜덤 시간만큼 슬립
            # 1초에서 3초 사이의 랜덤한 시간만큼 대기
            sleep_time = random.uniform(1, 2) 
            print(f'{year}년 {month}월 {week}주 {page}페이지 완료')
            print(f"{sleep_time:.2f}초 대기...")
            time.sleep(sleep_time)


# 데이터프레임으로 변환
df = pd.DataFrame(data_list)

df.reset_index(drop=True, inplace=True)
df
```

```python
df.to_csv('../data/24년_주차별_top500.csv', index=False)
```
