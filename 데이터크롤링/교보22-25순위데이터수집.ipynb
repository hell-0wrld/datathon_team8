```python
import requests
import pandas as pd
import time
import random

headers = {
    'User-Agent': (
    'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
    'AppleWebKit/537.36 (KHTML, like Gecko) '
    'Chrome/115.0.0.0 Safari/537.36'
    ),
}

months = ('01', '02', '03', '04', '05', '06', '07', '08', '09', '10', '11', '12')
data_list = []

for year in range(2022, 2026):
    for month in months:
        url = f'https://store.kyobobook.co.kr/api/gw/best/best-seller/total?page=1&per=200&period=003&bsslBksClstCode=A&ymw={year}{month}'
        response = requests.get(url, headers=headers)
        time.sleep(5)

        

        if response.status_code == 200:
            data = response.json()
            # 데이터가 있는 경우에만 처리
            if 'bestSeller' in data['data'] and data['data']['bestSeller']:
                for item in data['data']['bestSeller']: # 리스트의 각 항목을 직접 반복
                    # 순위
                    rank = item.get('prstRnkn')
                    # 순위 기준일
                    rank_date = item.get('ymw')
                    # 책 제목
                    name = item.get('cmdtName')
                    # 발행일
                    publ_date = item.get('rlseDate')
                    # ISBN
                    isbn = item.get('cmdtCode')
                    # 책 코드(교보)
                    kyob_code = item.get('saleCmdtid')
                    # 분류
                    cate = item.get('saleCmdtClstName')

                    data_list.append({
                        'rank': rank,
                        'rank_date': rank_date,
                        'name': name,
                        'publ_date': publ_date,
                        'isbn': isbn,
                        'kyob_code': kyob_code,
                        'cate': cate
                    })
            else:
                print(f"+++++{year}년 {month}월 없음+++++")
        else:
            print(f"====요청 실패: {response.status_code} ({year}년 {month}월)====")

        sleep_time = random.uniform(5, 6)
        print(f'{year}년 {month}월 수집 중')
        print(f"{sleep_time:.2f}초 대기...")
        time.sleep(sleep_time)

df = pd.DataFrame(data_list)
print(df.head())
```

```python
df.to_csv('교보문고_2125_2312월누락.csv', encoding='utf-8', index=False)
```
