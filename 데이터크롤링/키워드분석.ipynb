```python
# 4. 키워드 추출
keywords = summarize_with_keywords(
    reviews,
    min_count=2,     # 최소 n번 이상 등장하는 단어만
    max_length=10,   # 최대 글자 길이
    beta=0.85,       # 페이지랭크 계수(기본값)
    max_iter=10,     # 반복 횟수(기본값)
    verbose=True
)

print("추출된 키워드:")
for word, score in keywords.items():
    print(f"{word}: {score:.4f}")
```

```python
# 5. 워드클라우드 시각화
font_path = '/usr/share/fonts/truetype/nanum/NanumGothic.ttf'
wordcloud = WordCloud(
    font_path=font_path,
    width=800, height=500, background_color='white'
).generate_from_frequencies(keywords)

plt.figure(figsize=(10, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.rc('font', family='NanumGothic')  # 폰트 적용
plt.rcParams['axes.unicode_minus'] = False
font_path = '/usr/share/fonts/truetype/nanum/NanumGothic.ttf'

# 6. 데이터프레임 변환
df = pd.DataFrame(list(keywords.items()), columns=["단어", "점수"]).sort_values(by="점수", ascending=False)

# 7. 상위 15개 키워드 그래프 시각화
top_n = 15
df_top = df.head(top_n)

plt.figure(figsize=(10, 6))
plt.barh(df_top["단어"], df_top["점수"])
plt.gca().invert_yaxis()
plt.xlabel("점수")
plt.title(f"도서 리뷰 키워드 상위 {top_n}개")
plt.show()
```
